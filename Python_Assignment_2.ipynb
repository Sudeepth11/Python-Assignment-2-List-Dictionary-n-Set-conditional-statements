List (Creation, Modification and Access):  

```python
##--1. List Creation: 
age_list=[24,33,31,18,27,40]
name_list=["Sudeepth","Yash","Abhinav","Adhithyan","Anju"]
print(name_list)
print(age_list)

##--2. List Operations / Modifications: 
name_list.append("Yazhini")
age_list.insert(2,30)

name_list.remove("Yazhini")
age_list.pop()

age_list.extend([29,30,26])
age_list.sort(reverse=True)

##--3. Accessing List Elements: 
print(name_list[0])
print(name_list[-1])
print(name_list[2:4])
print(name_list[::-1])
```

Dictionary (Creation, Modification and Access):  


```python
student_marks = {
    "Sudeepth": 85,
    "Bob": 92,
    "Abhinav": 78,
    "Adhithyan": 88,
    "Ashlyn": 95
}

print(student_marks["Bob"])

student_marks["Janani"] = 80

student_marks["Bob"]=82

print(student_marks.keys())
print(student_marks.values())
print(student_marks.items())
```

Sets (Operations):  

```python
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}

print(my_set)
# Explanation
# ----set in Python is a collection of unique elements.
# ----Sets do not allow duplicate values.(In the given values, 'a' appears three times and 'i' appears twice,
#     but the extras are automatically removed once included as sets.Therefore, only five unique vowels are 
#     stored in my_set

my_set[4] = 's'
# Explanation
# ----set is an unordered collection of unique elements.
# ----Sets do not support indexing because their elements do not have fixed positions.
#     my_set[4] = 's' attempts to modify an element using its index.
#     Since sets do not support item assignment, Python raises a TypeError.

set1={1,3,5,7,9}
set2={2,3,5,8,10}

print("Union:",set1.union(set2))
print("Intersection:",set1.intersection(set2))
```

Operators & Conditional Statements : 
(IF, ELIF, ELSE) 

```python
score=int(input("Enter your score (0 to 10): "))
if score>7:
    print("Above Average: Great Performance ,Continue through same Efforts.")
elif score>=4 && score<=7:
    print("Average: Good effort! Keep practicing, there's room for improvement.")
else:
    print("")
```
